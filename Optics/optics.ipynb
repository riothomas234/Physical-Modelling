```python
import numpy as np
import scipy.integrate
import matplotlib.pyplot as plt

lambd = 500*(10**-9)

k=2*np.pi/lambd

d= 0.001

Y_s = np.linspace(-10,10,1000)


D_s = [0.01, 0.1, 0.326,0.64, 4]

def d_phi(y,Y):
    r= ((Y-y)**2+D**2)**(1/2)
    return np.exp(1j*k*r)




```

```python

for D in D_s:
    phi_s = np.array([
        scipy.integrate.quad(d_phi, -d/2, d/2, args=(Y,))[0]
        for Y in Y_s
    ])

    plt.plot(Y_s,abs(phi_s)**2 )
    plt.show()
```
